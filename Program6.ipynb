```python
import codecs
sourceFileName = "datasets/spam.csv"
targetFileName = "datasets/spam2.csv"
BLOCKSIZE = 1048576 # or some other, desired size in bytes
with codecs.open(sourceFileName, "r", "iso-8859-1") as sourceFile:
    with codecs.open(targetFileName, "w", "utf-8") as targetFile:
        while True:
            contents = sourceFile.read(BLOCKSIZE)
            if not contents:
                break
            targetFile.write(contents)
```

```python
sms_collection = pd.read_csv("datasets/spam2.csv", sep = ",")
```

```python
sms_collection = sms_collection.drop(columns = ["Unnamed: 2", "Unnamed: 3", "Unnamed: 4"])
```

```python
sms_collection = sms_collection.rename(columns = {"v1":"Label", "v2":"Message"})
```

```python
sms_collection
```

```python
stopwords_set = set(stopwords.words("english"))
```

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^0-9a-zA-Z]', ' ', text)
    text = re.sub(r'\s+', ' ', text)
    text = " ".join(word for word in text.split() if word not in stopwords_set)
    return text
```

```python
sms_collection["Message"] = sms_collection['Message'].apply(clean_text)
```

```python
sms_collection["Sentence_Length"] = sms_collection["Message"].apply(lambda x: len(x))
```

```python
pos = nltk.pos_tag(nltk.word_tokenize(sentence))
```

```python
pos_freq = Counter(tag for _, tag in pos)
```
