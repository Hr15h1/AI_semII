Create images with different non overlapped filled shapes. Implement K-means clustering for colour segmentation using those images. Display the shapes seperately

```python
import cv2
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import numpy as np
```

```python
shapes_img = cv2.imread("images/shapes.png")
converted_img = cv2.cvtColor(shapes_img, cv2.COLOR_BGR2RGB)
original_shape = shapes_img.shape

# plt.imshow(converted_img)

img_vector = converted_img.reshape((-1, 3))
print(f"The shape of the image after converting it into a vector: {img_vector.shape}")

# cluster_range = [i for i in range(8)]

# optimal_clusters = 0
k = 8
# for cluster in cluster_range:
km = KMeans(n_clusters=k)
cluster_labels = km.fit_predict(img_vector)
centers = km.cluster_centers_
centers = np.array(centers, dtype = 'uint8')

i=1
colors = []
plt.figure(figsize=(8,2))
for center in centers:

    plt.subplot(1, 8, i)

    
    plt.axis("off")
    i+=1

    colors.append(center)

    a = np.zeros((100, 100, 3), dtype = 'uint8')
    a[:, :, :] = center
    plt.imshow(a)


    # res = center[]

for i in range(k):
    img_vector_copy = img_vector.copy()
    img_vector_copy[cluster_labels != i] = 255
    img_vector_copy = img_vector_copy.reshape(original_shape)
    plt.figure(figsize=(5,5))
    plt.axis("off")
    plt.imshow(img_vector_copy)



```
