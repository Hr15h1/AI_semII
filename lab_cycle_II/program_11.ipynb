Implement a Python script to model a Markov Decision Process using a given transition matrix and reward function of 3 states and 2 actions. Calculate the value function for each state provided the policy for all states and actions are initialized as 0.5. Provide a sample MDP and demonstrate the corresponding value function (Set discount factor = 0.9). 

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
states = [0, 1, 2]
actions = [0, 1]
discount_factor = 0.9

policy = {0: {0: 0.5, 1: 0.5},
          1: {0: 0.5, 1: 0.5},
          2: {0: 0.5, 1: 0.5}}

transitions = {
    0: {
        0: [(0.6, 0, 5)],
        1: [(0.4, 2, 10)]
    },
    1: {
        0: [(0.5, 1, 3)],
        1: [(0.5, 2, 4)]
    },
    2: {
        0: [(0.7, 0, -5)],
        1: [(0.3, 1, 2)]
    }
}
```

```python
def polic_evaluation(states, actions, policy, transitions, discount_factor, theta=1e-6):
    V = {s: 0.0 for s in states}
    while True:
        delta = 0
        for s in states:
            v = 0
            for a in actions:
                action_prob = policy[s][a]
                for (prob, next_state, reward) in transitions[s][a]:
                    v += action_prob * prob * (reward + discount_factor * V[next_state])
            delta = max(delta, abs(v - V[s]))
            V[s] = v
        if delta < theta:
            break
    return V
```

```python
v = polic_evaluation(states, actions, policy, transitions, discount_factor)
print("State Values:", v)
```
