Implement a Python script to model a Markov Decision Process using a given transition matrix and reward function of 3 states and 2 actions. Calculate the value function for each state provided the policy for all states and actions are initialized as 0.5. Provide a sample MDP and demonstrate the corresponding value function (Set discount factor = 0.9). 

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
states = [0, 1, 2]
actions = [0, 1]
discount_factor = 0.9

policy = {0: {0: 0.5, 1: 0.5},
          1: {0: 0.5, 1: 0.5},
          2: {0: 0.5, 1: 0.5}}

transitions = {
    0: {
        0: [(0.6, 0, 5)],
        1: [(0.4, 2, 10)]
    },
    1: {
        0: [(0.5, 1, 3)],
        1: [(0.5, 2, 4)]
    },
    2: {
        0: [(0.7, 0, -5)],
        1: [(0.3, 1, 2)]
    }
}
```

```python
def polic_evaluation(states, actions, policy, transitions, discount_factor, theta=1e-6):
    V = {s: 0.0 for s in states}
    while True:
        delta = 0
        for s in states:
            v = 0
            for a in actions:
                action_prob = policy[s][a]
                for (prob, next_state, reward) in transitions[s][a]:
                    v += action_prob * prob * (reward + discount_factor * V[next_state])
            delta = max(delta, abs(v - V[s]))
            V[s] = v
        if delta < theta:
            break
    return V
```

```python
v = polic_evaluation(states, actions, policy, transitions, discount_factor)
```

```python
print("The value function for each state is:")
for state in states:
    print(f"State {state}: {v[state]:.2f}") 
```

```python
# Define parameters
num_states = 3
num_actions = 2
gamma = 0.9  # Discount factor

# Sample transition matrix: T[s][a][s'] = P(s' | s, a)
T = np.array([
    [[0.8, 0.2, 0.0], [0.1, 0.6, 0.3]],  # Transitions from state 0
    [[0.0, 0.9, 0.1], [0.4, 0.0, 0.6]],  # Transitions from state 1
    [[0.2, 0.4, 0.4], [0.3, 0.3, 0.4]]   # Transitions from state 2
])

# Reward function: R[s][a]
R = np.array([
    [5, 10],    # Rewards for actions in state 0
    [-1, 2],    # Rewards for actions in state 1
    [0, 4]      # Rewards for actions in state 2
])

# Policy: π[s][a] = probability of taking action a in state s
# Initialized to uniform (0.5 for each action)
policy = np.full((num_states, num_actions), 0.5)

# Value function: V[s]
V = np.zeros(num_states)

# Iterative policy evaluation
def evaluate_policy(policy, T, R, gamma, theta=1e-6):
    V = np.zeros(num_states)
    while True:
        delta = 0
        for s in range(num_states):
            v = 0
            for a in range(num_actions):
                action_prob = policy[s][a]
                reward = R[s][a]
                transition_probs = T[s][a]
                v += action_prob * (reward + gamma * np.dot(transition_probs, V))
            delta = max(delta, abs(v - V[s]))
            V[s] = v
        if delta < theta:
            break
    return V

# Evaluate the initialized policy
V = evaluate_policy(policy, T, R, gamma)

# Display the results
for s in range(num_states):
    print(f"Value of state {s}: {V[s]:.4f}")
```
