Implement and compare defuzzification methods using the fuzzy output set. {10: 0.0, 20: 0.3, 30: 0.5, 40: 0.8, 50: 1.0, 60: 0.8, 70: 0.4, 80: 0.2}. Visualize the fuzzy output and highlight each defuzzified result on the plot.

```python
import skfuzzy as fuzzy
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.arange(10, 90, 10)
```

```python
y = np.array([0.0, 0.3, 0.5, 0.8, 1.0, 0.8, 0.4, 0.2])
```

We can defuzzify the above fuzzy set in 6 ways

1. Centroid of Area (centroid)
2. Bisector of Area (bisector)
3. Mean of Maximum (mom)
4. Min of Maximum (som)
5. Max of Maximum (lom)

```python
methods = ['centroid', 'bisector', 'mom', 'som', 'lom']
colors = ['r', 'g', 'b', 'c', 'm']

plt.figure(figsize=(10, 6))
for method, color in zip(methods, colors):
    center = fuzzy.defuzz(x, y, method)
    print(f"Defuzzification method: {method}, Center: {center:.2f}")
    plt.plot(x, y, color=color, label=f'{method} ({center:.2f})')
    plt.axvline(center, color=color, linestyle='--')
plt.title('Fuzzy Defuzzification Methods')
plt.xlabel('Input')
plt.ylabel('Output')
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import skfuzzy as fuzz

# Step 1: Define the fuzzy output set
x = np.array([10, 20, 30, 40, 50, 60, 70, 80])
membership = np.array([0.0, 0.3, 0.5, 0.8, 1.0, 0.8, 0.4, 0.2])

# Step 2: Interpolate to a finer universe for better defuzzification
x_fine = np.linspace(10, 80, 1000)
membership_fine = np.interp(x_fine, x, membership)

# Step 3: Apply defuzzification methods
centroid = fuzz.defuzz(x_fine, membership_fine, 'centroid')
bisector = fuzz.defuzz(x_fine, membership_fine, 'bisector')
mom = fuzz.defuzz(x_fine, membership_fine, 'mom')
som = fuzz.defuzz(x_fine, membership_fine, 'som')
lom = fuzz.defuzz(x_fine, membership_fine, 'lom')

# Step 4: Print results
print("Defuzzification Results:")
print(f"Centroid: {centroid:.2f}")
print(f"Bisector: {bisector:.2f}")
print(f"Mean of Maximum (MoM): {mom:.2f}")
print(f"Smallest of Maximum (SoM): {som:.2f}")
print(f"Largest of Maximum (LoM): {lom:.2f}")

# Step 5: Visualization
plt.figure(figsize=(12, 6))
plt.plot(x, membership, 'bo', label='Fuzzy Output Points')
plt.plot(x_fine, membership_fine, 'b-', alpha=0.6, label='Interpolated Fuzzy Set')

# Plot defuzzified values
plt.axvline(centroid, color='r', linestyle='--', label=f'Centroid: {centroid:.2f}')
plt.axvline(bisector, color='g', linestyle='--', label=f'Bisector: {bisector:.2f}')
plt.axvline(mom, color='purple', linestyle='--', label=f'MoM: {mom:.2f}')
plt.axvline(som, color='orange', linestyle='--', label=f'SoM: {som:.2f}')
plt.axvline(lom, color='brown', linestyle='--', label=f'LoM: {lom:.2f}')

plt.title("Defuzzification of Fuzzy Output Set")
plt.xlabel("Output Variable (e.g., Temperature)")
plt.ylabel("Membership Degree")
plt.grid(True)
plt.legend()
plt.show()
```
