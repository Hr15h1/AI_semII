Write a Python program to fuzzify the input variable temperature using Triangular membership functions for Low, Medium, and High with Input temperature values [10, 20, 30, 40, 50].

```python
import skfuzzy as fuzzy
import numpy as np
import matplotlib.pyplot as plt
```

```python
temperature = np.array([10, 20, 30, 40, 50])

temps = np.linspace(10, 50, 100)
```

```python
low = fuzzy.trimf(temps, [10, 10, 30])
medium = fuzzy.trimf(temps, [20, 30, 40])
high = fuzzy.trimf(temps, [30, 50, 50])
```

```python
plt.figure(figsize=(10, 5))
plt.plot(temps, low, label='Low', color='blue')
plt.plot(temps, medium, label='Medium', color='green')
plt.plot(temps, high, label='High', color='red')
plt.title('Fuzzy Membership Functions for Temperature')
plt.xlabel('Temperature (°C)')
plt.ylabel('Membership Degree')
plt.legend()
plt.grid()
plt.show()
```

```python
for temp in temperature:
    chance_low = fuzzy.interp_membership(temps, low, temp)
    chance_medium = fuzzy.interp_membership(temps, medium, temp)
    chance_high = fuzzy.interp_membership(temps, high, temp)

    chances = {
        'Low': chance_low,
        'Medium': chance_medium,
        'High': chance_high
        }
    
    max_chance = max(chances, key=chances.get)

    print(f"Temperature {temp}°C: is a {max_chance}")
    


```

```python
import numpy as np
import skfuzzy as fuzz
import matplotlib.pyplot as plt

# Step 1: Define temperature universe
temperature = np.array([10, 20, 30, 40, 50])

# Step 2: Create a finer universe for smooth plotting and calculation
x_temp = np.linspace(10, 50, 100)

# Step 3: Define triangular membership functions
low = fuzz.trimf(x_temp, [10, 10, 30])     # Peak at 10, falls off by 30
medium = fuzz.trimf(x_temp, [20, 30, 40])  # Symmetric around 30
high = fuzz.trimf(x_temp, [30, 50, 50])    # Rises from 30, peak at 50

# Step 4: Fuzzify the input temperature values
print("Fuzzified outputs:")
for temp in temperature:
    low_val = fuzz.interp_membership(x_temp, low, temp)
    med_val = fuzz.interp_membership(x_temp, medium, temp)
    high_val = fuzz.interp_membership(x_temp, high, temp)
    print(f"Temperature: {temp} -> Low: {low_val:.2f}, Medium: {med_val:.2f}, High: {high_val:.2f}")

# Step 5: Plotting the membership functions
plt.figure(figsize=(10, 5))
plt.plot(x_temp, low, label='Low', color='blue')
plt.plot(x_temp, medium, label='Medium', color='green')
plt.plot(x_temp, high, label='High', color='red')

# Plot the input points on the curves
for temp in temperature:
    plt.axvline(temp, color='gray', linestyle='--', alpha=0.3)

plt.title("Triangular Membership Functions for Temperature")
plt.xlabel("Temperature")
plt.ylabel("Membership Degree")
plt.legend()
plt.grid(True)
plt.show()
 
```
