Write a Python program to fuzzify the input variable temperature using Triangular membership functions for Low, Medium, and High with Input temperature values [10, 20, 30, 40, 50].

```python
import skfuzzy as fuzzy
import numpy as np
import matplotlib.pyplot as plt
```

```python
temperature = np.array([10, 20, 30, 40, 50])
```

```python
low = fuzzy.trimf(temperature, [10, 20, 30])
medium = fuzzy.trimf(temperature, [20, 30, 40])
high = fuzzy.trimf(temperature, [30, 40, 50])
```

```python
plt.figure(figsize=(10, 5))
plt.plot(temperature, low, label='Low', color='blue')
plt.plot(temperature, medium, label='Medium', color='green')
plt.plot(temperature, high, label='High', color='red')
plt.title('Fuzzy Membership Functions for Temperature')
plt.xlabel('Temperature (°C)')
plt.ylabel('Membership Degree')
plt.legend()
plt.grid()
plt.show()
```
