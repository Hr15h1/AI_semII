Bell-shaped Membership Function

```python
set_a = np.linspace(0, 10, 100)

width = 2
slope = 2.5
center = 5

m_a = 1 / (1 + np.abs((set_a - center) / width) ** slope)

plt.plot(set_a, m_a, 'g', label = 'Bell-shaped Membership Function')
plt.xlabel('Universe of Discourse')
plt.ylabel('Membership Degree')
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Membership Functions ---

def triangular(x, a, b, c):
    """Triangular membership function with feet at a and c, peak at b."""
    return np.maximum(np.minimum((x - a) / (b - a), (c - x) / (c - b)), 0)

def trapezoidal(x, a, b, c, d):
    """Trapezoidal membership function with slopes from a to b and c to d."""
    return np.maximum(np.minimum(np.minimum((x - a)/(b - a), 1), (d - x)/(d - c)), 0)

def bell_shaped(x, a, b, c):
    """Generalized bell-shaped membership function centered at c."""
    return 1 / (1 + np.abs((x - c) / a) ** (2 * b))

# --- Universe of Discourse ---
x = np.linspace(0, 10, 500)

# --- Membership Function Values ---
tri_y = triangular(x, 2, 5, 8)
trap_y = trapezoidal(x, 2, 4, 6, 8)
bell_y = bell_shaped(x, a=1, b=4, c=5)

# --- Plotting ---
plt.figure(figsize=(10, 6))

plt.plot(x, tri_y, label="Triangular (a=2, b=5, c=8)", color='blue')
plt.plot(x, trap_y, label="Trapezoidal (a=2, b=4, c=6, d=8)", color='green')
plt.plot(x, bell_y, label="Bell-shaped (a=1, b=4, c=5)", color='red')

plt.title("Triangular, Trapezoidal, and Bell-Shaped Membership Functions")
plt.xlabel("x")
plt.ylabel("Membership Degree")
plt.ylim(-0.1, 1.1)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
