Write a Python program to plot Triangular, Trapezoidal, and Bell-shaped membership functions using NumPy and Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
set_a = np.linspace(0, 10, 5)
```

```python
set_a
```

Triangular Membership Function

```python
y = [0.1, 0.5, 0.9, 0.5, 0.1]
plt.plot(set_a, y, 'r', label = 'Fuzzy Set A')
```

Trapezoidal Membership Function

```python
y = [0.1, 0.5, 0.5, 0.5, 0.1]
plt.plot(set_a, y, 'b', label = 'Fuzzy Set B')
```

Bell-shaped Membership Function

```python
set_a = np.linspace(0, 10, 100)

width = 2
slope = 2.5
center = 5

m_a = 1 / (1 + np.abs((set_a - center) / width) ** slope)

plt.plot(set_a, m_a, 'g', label = 'Bell-shaped Membership Function')
plt.xlabel('Universe of Discourse')
plt.ylabel('Membership Degree')
```
